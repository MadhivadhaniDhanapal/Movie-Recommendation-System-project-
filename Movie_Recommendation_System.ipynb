```
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Load Datasets
movies = pd.read_csv("ml-latest-small/movies.csv")
ratings = pd.read_csv("ml-latest-small/ratings.csv")

print("Datasets Loaded Successfully!")

# Merge Datasets
movie_data = pd.merge(ratings, movies, on="movieId")

# Create User-Movie Matrix
movie_matrix = movie_data.pivot_table(
    index="title",
    columns="userId",
    values="rating"
).fillna(0)

print("User-Movie Matrix Created!")

# Calculate Cosine Similarity
similarity = cosine_similarity(movie_matrix)

# Create Similarity DataFrame
similarity_df = pd.DataFrame(
    similarity,
    index=movie_matrix.index,
    columns=movie_matrix.index
)

print("Similarity Matrix Created Successfully!")

# Recommendation Function
def recommend(movie_name, n=5):
    if movie_name not in similarity_df.columns:
        print("Movie not found!")
        return

    similar_movies = similarity_df[movie_name].sort_values(
        ascending=False
    )[1:n+1]

    print("\nTop Recommended Movies:\n")

    for movie in similar_movies.index:
        print(movie)

# User Input
movie = input("Enter Movie Name: ")
recommend(movie)
```

```
# ==========================================
# Movie Recommendation System
# Phase 1 - Recommendation Engine
# ==========================================

import pandas as pd

# Step 1: Load Dataset
movies = pd.read_csv("ml-latest-small/movies.csv")
ratings = pd.read_csv("ml-latest-small/ratings.csv")

print("✅ Dataset Loaded Successfully")
print("Movies :", len(movies))
print("Ratings :", len(ratings))

# Step 2: Merge Datasets
movie_data = pd.merge(movies, ratings, on="movieId")

# Step 3: Calculate Average Rating
movie_summary = movie_data.groupby(["movieId","title"]).agg(
    Average_Rating=("rating","mean"),
    Total_Ratings=("rating","count")
).reset_index()

# Step 4: Filter Popular Movies
popular_movies = movie_summary[movie_summary["Total_Ratings"] >= 100]

# Step 5: Sort by Rating
recommended_movies = popular_movies.sort_values(
    by=["Average_Rating","Total_Ratings"],
    ascending=False
).head(5)

# Step 6: Display Recommendation

print("\n")
print("="*50)
print("🎬 TOP RECOMMENDED MOVIES")
print("="*50)

for i,row in enumerate(recommended_movies.itertuples(),start=1):

    print(f"\n{i}. {row.title}")
    print(f"⭐ Rating : {row.Average_Rating:.2f}/5")
    print(f"👥 Users Rated : {row.Total_Ratings}")

print("\n✅ Recommendation Completed")
```
